```python
import rocksdbpy
import json
```

```python
db = rocksdbpy.open_default('/home/infres/rapaport/datasets/2025-05-28-popular-1k/results/RepositoryStats.db')
```

```python
for key, value in db.iterator():
    print("Key:", key)

    value_str = value.decode('utf-8')
    value_json = json.loads(value_str)
    
    print("Value as JSON:", value_json)
    print(type(value_json['commit_time']))
```

```python
db.close()
```

```python
value_json['commit_time'][0]
```

```python
time_between = [value_json['commit_time'][i+1] - value_json['commit_time'][i] for i in range(len(value_json['commit_time'])-1)]
```

```python
mean_time_between = sum(time_between)/len(time_between)
```

```python
print(f'mean time between commits : {mean_time_between/3600}h')
```
